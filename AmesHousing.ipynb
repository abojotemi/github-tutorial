### Practical Notes for Modeling

- Many `Qual`/`Cond` columns are **ordinal** (Excellent/Good/Average/Fair/Poor). Encode these as ordered integers rather than one-hot — it typically improves model performance over treating them as unordered categories.
- Watch for `NaN` in columns like `Alley`, `Pool QC`, `Fence`, and `Misc Feature`. These usually mean the house **doesn't have that feature**, not that data is missing — don't drop those rows, impute them with a meaningful category like `"None"` instead.
